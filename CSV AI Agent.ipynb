```python
import os
import pandas as pd
from langchain_google_genai import ChatGoogleGenerativeAI  #import the gemini ai to the code
from langchain_experimental.agents import create_pandas_dataframe_agent  #lets and ai model connnect to a dataframe

os.environ['GOOGLE_API_KEY'] = 'API KEY'

df = pd.read_csv('/Workspace/Users/kushal.vadukul2006@gmail.com/LangChain-CSV-Agent/bank_transactions_data_2.csv')

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0)

agent_chain = create_pandas_dataframe_agent(
    llm,
    df,
    allow_dangerous_code=True,  #lets langchain know that i allowed the ai to write code.
    agent_executor_kwargs={"handle_parsing_errors": True}  #if code gives error dont stop code, just correct the error and go again.
)

user_input = input("Ask a question about the Dataset: ")   #input from user

response = agent_chain.invoke(user_input)

print(response) #response from ai

with open("csv_agent_answers.txt", "a") as f:  #writes the question and answer into a txt file in same folder called csv_agent_answers.
    f.write(f"Question: {user_input}\n")
    f.write(f"Answer:\n{response}\n")
```
